```python
def rakna_annonser(antal):
    resultat = antal + 1
    return resultat
antal_annonser = rakna_annonser(antal_annonser)

print(antal_annonser)
```

```python
def finns_teknik(text, teknik):
    if teknik.lower() in text.lower():
        return True
    else:
        return False

annons = "vi söker en AI-utvecklare med kunskaper i Python och SQL"
resultat = finns_teknik(annons, "python")

print("Resultatet är:", resultat)
```

```python
jobannoser = [
["Ai-utvecklare", "Python", "SQL"],
["Machine Learning Engineer", "Python", "AWS"],
["Data Scientist", "Python", "SQL"],
["AI Enginer", "Python", "Docker"],
["Data Enginer", "SQL", "AWS"]
]
```

```python
python_antal = 0

for jobb in jobannoser:
    if "Python" in jobb:
     python_antal = python_antal + 1

print("Antal jobb som kräver python", python_antal)
```

```python
sql_antal = 0

for jobb in jobannoser:
    if "SQL" in jobb:
        sql_antal = sql_antal + 1
print("Antal jobb som kräver SQL:", sql_antal)
```

```python
aws_antal = 0
for jobb in jobannoser:
    if "AWS" in jobb:
        aws_antal = aws_antal + 1
print("Antal jobb som krävs AWS:", aws_antal)
```

```python
docker_antal = 0
for jobb in jobannoser:
    if "Docker" in jobb:
        docker_antal = docker_antal + 1 
print("Antal jobb som kräver Docker:", docker_antal)
```

```python
tekniker = ["Python", "SQL", "AWS", "Docker"]

for teknik in tekniker:
    antal = 0

    for jobb in jobannoser:
        if teknik in jobb:
            antal = antal + 1
    print(teknik, "finns i", antal, "jobb")
```

```python
resultat = {}
for teknik in tekniker:
    antal = 0
    for jobb in jobannoser:
        if teknik in jobb:
            antal = antal + 1
    resultat[teknik] = antal
print(resultat)
```

```python
mest_efterfrågad = max(resultat, key=resultat.get)
print("Mest efterfrågade tekniken är:", mest_efterfrågad)
print("Antal jobb:", resultat[mest_efterfrågad])
```

```python
import matplotlib.pyplot as plt 

tekniker = list(resultat.keys())
antal = list(resultat.values())

plt.bar(tekniker, antal)

plt.title("antal jobbannonser per teknik")
plt.xlabel("Teknik")
plt.ylabel("Antal jobb")
```
