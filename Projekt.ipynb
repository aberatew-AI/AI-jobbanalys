```python
from groq import Groq

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

print("Groq är ansluten")
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "user",
            "content": "Säg hej och förklara kort vad en AI-utvecklare gör."
        }
    ]
)

print(response.choices[0].message.content)
```

```python
tekniker = {
    "Python": 4,
    "SQL": 3,
    "AWS": 2,
    "Docker": 1
}

print(tekniker)
```

```python
import json 

with open("jobbdata.json", "w", encoding="utf-8") as fil:
    json.dump(tekniker, fil, ensure_ascii=False, indent=4)

print("Data sparad i jobbdata.json")
```

```python
prompt = f"""
Här är data från jobbannonser för AI-utvecklare:

{tekniker}

Analysera datan och förklara kort:
1. Vilken teknik är mest efterfrågad?
2. Vilken är minst efterfrågad?
3. Vad borde en student prioritera att lära sig?
"""

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ]
)

print(response.choices[0].message.content)
```

```python
ai_analys = response.choices[0].message.content

print("AI-analys av jobbannonser:")
print(ai_analys)
```

```python
plt.bar(tekniker.keys(), tekniker.values())

plt.title("Efterfrågade tekniker för AI-utvecklare")
plt.xlabel("Teknik")
plt.ylabel("Antal jobbannonser")

plt.show()
```
